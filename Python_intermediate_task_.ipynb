```python
total_products = int(input("Enter total number of products: "))
total_bill = 0
for i in range(1, total_products + 1):
    product_name = input(f"Enter the name of product {i}: ")
    product_price = int(input("Enter the price of the product: "))
    product_quantity = int(input("Enter the quantity of the product: "))

    sub_total = product_price * product_quantity
    total_bill = total_bill + sub_total
if total_bill >= 5000:
    discount = (total_bill * 10) / 100
else:
    discount = 0

final_bill = total_bill - discount
print("========== Bill Summary ==========")
print("Total Bill:", total_bill)
print("Discount:", discount)
print("Final Bill:", final_bill)
```

```python
employee_name = str(input("Enter Employee Name: "))
hourly_rate = float(input("Enter Hourly Rate: "))
total_Whrs = float(input("Enter Total Working Hours: "))
if total_Whrs <=40:
  normal_salary = hourly_rate * total_Whrs
  overtime_salary = 0
else :
  normal_salary = hourly_rate * 40
  overtime_hours = total_Whrs - 40
  overtime_rate = hourly_rate * 1.5
  overtime_salary = overtime_hours * overtime_rate

  total_salary = normal_salary + overtime_salary
print("========== Employee Salary ==========")
print("Employee Name:", employee_name)
print("Hourly Rate:", hourly_rate)
print("Total Working Hours:", total_Whrs)
print("Total Salary:", total_salary)
```

```python
message = input("Enter your message: ").title()

total_character = len(message)

total_words = len(message.split())

vowels = "aeiouAEIOU"
vowels_count = 0
for vowel in message:
  if vowel in vowels:
    vowels_count += 1

total_spaces = message.count(" ")

upper_case_ = message.upper()

if "python" or "PYTHON" in message:
  python_exist = "yes"
else:
  python_exist = "no"

print("========== Message Analysis ==========")


print("Message: ", message)
print("Total character: ", total_character)
print("Total words: ", total_words)
print("Total vowel: ", vowels_count)
print("Total spaces: ", total_spaces)
print("Uppercase_formating: ", upper_case_ )
print("Python Exist: ", python_exist)
```

```python
contact_info={}
while True :
  print("1.Add Contact")
  print("2.Search Contact")
  print("3.Display All Contact")
  print("4.Exit")
  choice = int(input("Enter Your Choice 1-4:"))

  if choice==1:
    name=input("Enter Contact Name:")
    number=input("Enter Contact Number:")
    if name in contact_info:
      print("Contact Already Exists")
    else:
      contact_info[name]=number
      print("Contact Added")
  elif choice==2:
    name=input("Enter Contact Name:")
    if name in contact_info:
      print("Contact Found:", contact_info[name])
    else:
      print("Contact Not Found")
  elif choice == 3:
   print("All Contacts:", contact_info)
  elif choice == 4:
   print("Thank You")
   break
  else:
   print("Invalid Choice")
```
