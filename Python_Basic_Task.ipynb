```python
#Task 1
user_name = (input("Enter Your Name: "))
user_age = int(input("Enter Your Age: "))

if user_age < 5 or user_age > 100:
  Price = "free"
elif user_age >= 5 and user_age <= 12:
   Price = 300
elif user_age  >= 13 and user_age <= 59:
   Price = 700
elif user_age  >= 60 and user_age <= 100:
   Price = 400

print ("Customer:" , user_name)
print ("Price:" , Price)
```

```python
#Task 2
secret_pin = 1804
for i in range(1, 4):
  user_pin = int(input("Enter Your Pin: "))
  if secret_pin == user_pin:
      print("Access Granted")
      break
  else:
      if i < 3:
        print("Try Again")
        print(f"No of attempts remaining: {3 - i}")
      else:
        print("Account Blocked")
```

```python
#Task 3
total_steps = 0
for i in range(1,8):
 steps = int(input(f"Enter Your day {i} Steps; "))
 total_steps = total_steps + steps

avg_st = total_steps / 7
print("Total Steps:", total_steps)
print("Average Steps:", avg_st)

if avg_st >= 5000:
    print("Active Week")
else:
    print("Needs More Activity")
```

```python
#Task 4
Temp_celsius = float(input("Enter Celsius Temp: "))
Fahrenheit = ( Temp_celsius * 9/5 + 32)
if Temp_celsius < 10:
  Weather = "Very Cold"
elif Temp_celsius >= 10 and Temp_celsius <= 19:
  Weather = "Cold"
elif Temp_celsius >= 20 and Temp_celsius <=29:
  Weather = "Normal"
else:
   Weather = "Hot"

print ("Fahrenheit_Temp:", Fahrenheit)
print ("Weather Condition:", Weather)

```
